
**Mantenimiento Predictivo (PdM)**

Es donde las máquinas se miden con métodos de tendencia monitoreada objetivamente, como el análisis de vibraciones y la imagen térmica infrarroja. El PdM es una herramienta valiosa ya que no requiere desmantelar una máquina para conocer su condición. El PdM define métodos para predecir o diagnosticar problemas en un equipo basado en la tendencia de los resultados de las pruebas. Estos métodos utilizan técnicas de prueba no intrusivas para medir y calcular las tendencias de rendimiento del equipo.

**El Mantenimiento Basado en Condición (CBM) es una metodología que combina el mantenimiento predictivo y preventivo con el monitoreo en tiempo real.**

El PdM utiliza sistemas de Mantenimiento Basado en Condición para detectar fuentes de fallos mucho antes de que ocurra una falla, haciendo del mantenimiento un proceso proactivo. El CBM detecta con precisión el estado actual de los sistemas mecánicos y predice la capacidad de los sistemas para funcionar sin fallas. Utiliza los niveles de falla potencial creados durante el proceso de diseño de la maquinaria, mide parámetros adecuados para cuantificar los niveles de falla existentes y ajusta el entorno operativo para hacer que estos niveles sean compatibles con la producción económica frente a la vida útil del equipo.

```python
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from google.colab import files
import io

# Function to upload and process the dataset
def load_data():
    uploaded = files.upload()
    for filename in uploaded.keys():
        data = pd.read_csv(io.BytesIO(uploaded[filename]))
        break  # Taking the first uploaded file
    return data

# Class for loading and cleaning data
class DataProcessor:
    def __init__(self, dataframe):
        self.df = dataframe

    def clean_data(self):
        # Handling missing values by dropping rows with any NaNs
        self.df = self.df.dropna()
        return self.df

    def get_features(self):
        # Selecting specific columns as features
        return self.df.iloc[:, [0, 1, 2, 3, 4]].values

# Class for performing KMeans clustering
class KMeansProcessor:
    def __init__(self, data):
        self.data = data

    def find_optimal_clusters(self):
        wcss = []
        # Finding within-cluster sum of squares for different number of clusters
        for i in range(1, 11):
            kmeans = KMeans(n_clusters=i, init='k-means++', random_state=0)
            kmeans.fit(self.data)
            wcss.append(kmeans.inertia_)
        return wcss

    def perform_clustering(self, num_clusters):
        # Performing KMeans clustering with specified number of clusters
        kmeans = KMeans(n_clusters=num_clusters, init='k-means++', random_state=0)
        self.kmeans_model = kmeans.fit(self.data)
        return self.kmeans_model.labels_

# Class for visualization operations
class Visualization:
    @staticmethod
    def scatter_plot(x, y, kmeans):
        # Scatter plot of clustered data points and centroids
        plt.figure(figsize=(8, 6))
        plt.scatter(x[y == 0, 0], x[y == 0, 1], s=50, c='yellow', label='C1')
        plt.scatter(x[y == 1, 0], x[y == 1, 1], s=50, c='blue', label='C2')
        plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=100, c='green', label='Centroid')
        plt.legend()
        plt.title('KMeans Clustering')
        plt.xlabel('Feature 1')
        plt.ylabel('Feature 2')
        plt.grid(True)
        plt.show()

    @staticmethod
    def heatmap(df):
        # Heatmap visualization of correlation matrix
        plt.figure(figsize=(10, 8))
        sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=.5)
        plt.title('Correlation Heatmap')
        plt.show()

# Class for RandomForestClassifier operations
class RandomForestProcessor:
    def __init__(self, x, y):
        # Initializing RandomForestClassifier
        self.classifier = RandomForestClassifier(n_estimators=10, criterion="entropy", random_state=0)
        self.classifier.fit(x, y)

    def predict(self, test):
        # Predicting using trained classifier
        return self.classifier.predict(test)

    def accuracy(self, y, y_pred):
        # Calculating accuracy score
        return accuracy_score(y, y_pred)

def main():
    # Load and clean data
    data = load_data()
    data_processor = DataProcessor(data)
    cleaned_data = data_processor.clean_data()
    features = data_processor.get_features()

    # Initialize KMeansProcessor for clustering
    kmeans_processor = KMeansProcessor(features)
    wcss = kmeans_processor.find_optimal_clusters()
    num_clusters = 2
    y_clusters = kmeans_processor.perform_clustering(num_clusters)

    # Visualize clustered data
    Visualization.scatter_plot(features, y_clusters, kmeans_processor.kmeans_model)

    # Visualize correlation heatmap of cleaned data
    Visualization.heatmap(cleaned_data)

    # Initialize RandomForestProcessor for classification
    random_forest = RandomForestProcessor(features, y_clusters)

    # Test data for prediction
    test_data = np.array([[5.1, 229, 65.23, 68, 0]])

    # Predict using RandomForestClassifier
    prediction = random_forest.predict(test_data)
    accuracy = random_forest.accuracy(y_clusters, random_forest.predict(features))
    print("Prediction:", prediction)
    print("Accuracy:", accuracy)

    # Function to predict based on classifier
    def give_pred(test):
        prediction = random_forest.predict(test)
        if prediction != 0:
            return 'Your System Works'
        return 'Your System Failed'

    # Test prediction function
    print(give_pred(test_data))

if __name__ == "__main__":
    main()
```

### Paso 3: Interpretación:

El clustering K-means es uno de los algoritmos de aprendizaje automático no supervisado más simples y populares.

Típicamente, los algoritmos no supervisados hacen inferencias a partir de conjuntos de datos utilizando solo vectores de entrada sin referirse a resultados conocidos o etiquetados.

El objetivo de K-means es simple: agrupar puntos de datos similares y descubrir patrones subyacentes. Para lograr este objetivo, K-means busca un número fijo (k) de clusters en un conjunto de datos.

### ¿Qué es un Mapa de Calor de Correlación?

Un mapa de calor de correlación es una herramienta gráfica que muestra la correlación entre múltiples variables como una matriz codificada por colores.

Es como una carta de colores 🌈 que nos muestra cuán relacionadas están diferentes variables.

En un mapa de calor de correlación, cada variable está representada por una fila y una columna, y las celdas muestran la correlación entre ellas.

El color de cada celda representa la fuerza y la dirección de la correlación, con colores más oscuros que indican correlaciones más fuertes.

### ¿Cómo Leer un Mapa de Calor de Correlación?

Observa el color de cada celda para ver la fuerza y la dirección de la correlación. Los colores más oscuros indican correlaciones más fuertes, mientras que los colores más claros indican correlaciones más débiles. Las correlaciones positivas (cuando una variable aumenta, la otra tiende a aumentar) suelen estar representadas por colores cálidos, como el rojo o el naranja. Las correlaciones negativas (cuando una variable aumenta, la otra tiende a disminuir) suelen estar representadas por colores fríos, como el azul o el verde. 📊

Entender los mapas de calor de correlación puede ayudarnos a identificar patrones y relaciones entre múltiples variables. Así que, la próxima vez que analices datos con muchas variables, ¡piensa como un artista y usa un mapa de calor de correlación para ver los colores de las relaciones!

```python
import numpy as np
import pandas as pd
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
from ipywidgets import interact, IntSlider, FileUpload
import io

# Clase para generar y limpiar datos simulados
class DataSimulator:
    def __init__(self, df=None):
        if df is not None:
            self.df = df
        else:
            self.df = self.generate_data()

    def generate_data(self):
        # Generar datos simulados
        np.random.seed(0)
        data = np.random.rand(500, 5) * 100
        columns = ['Feature1', 'Feature2', 'Feature3', 'Feature4', 'Feature5']
        df = pd.DataFrame(data, columns=columns)
        return df

    def clean_data(self):
        # Eliminar filas específicas con valores nulos (simulado)
        try:
            self.df = self.df.drop([199, 200, 201, 202], errors='ignore')
        except KeyError as e:
            print(f"KeyError: {e}")
        return self.df

    def get_features(self):
        # Seleccionar columnas específicas como características
        return self.df.iloc[:, [0, 1, 2, 3, 4]].values

# Clase para realizar clustering con KMeans
class KMeansProcessor:
    def __init__(self, data):
        self.data = data

    def find_optimal_clusters(self):
        wcss = []
        # Encontrar la suma de cuadrados dentro de los clusters para diferentes números de clusters
        for i in range(1, 11):
            kmeans = KMeans(n_clusters=i, init='k-means++', random_state=0)
            kmeans.fit(self.data)
            wcss.append(kmeans.inertia_)
        return wcss

    def perform_clustering(self, num_clusters):
        # Realizar clustering con KMeans con el número especificado de clusters
        kmeans = KMeans(n_clusters=num_clusters, init='k-means++', random_state=0)
        self.kmeans_model = kmeans.fit(self.data)
        return self.kmeans_model.labels_

# Clase para operaciones de visualización
class Visualization:
    @staticmethod
    def scatter_plot(x, y, kmeans, download=False):
        # Gráfico de dispersión de puntos de datos agrupados y centroides
        plt.figure(figsize=(8, 6))
        plt.scatter(x[y == 0, 0], x[y == 0, 1], s=50, c='yellow', label='C1')
        plt.scatter(x[y == 1, 0], x[y == 1, 1], s=50, c='blue', label='C2')
        plt.scatter(kmeans.cluster_centers_[:, 0], kmeans.cluster_centers_[:, 1], s=100, c='green', label='Centroid')
        plt.legend()
        plt.title('KMeans Clustering')
        plt.xlabel('Feature 1')
        plt.ylabel('Feature 2')
        plt.grid(True)

        if download:
            plt.savefig('/content/kmeans_scatter_plot.png')
            print("Imagen guardada como 'kmeans_scatter_plot.png' en /content")

        plt.show()

    @staticmethod
    def heatmap(df, download=False):
        # Visualización de un mapa de calor de la matriz de correlación
        plt.figure(figsize=(10, 8))
        sns.heatmap(df.corr(), annot=True, cmap='coolwarm', fmt='.2f', linewidths=.5)
        plt.title('Correlation Heatmap')

        if download:
            plt.savefig('/content/correlation_heatmap.png')
            print("Imagen guardada como 'correlation_heatmap.png' en /content")

        plt.show()

# Clase para operaciones de RandomForestClassifier
class RandomForestProcessor:
    def __init__(self, x, y):
        # Inicializar RandomForestClassifier
        self.classifier = RandomForestClassifier(n_estimators=10, criterion="entropy", random_state=0)
        self.classifier.fit(x, y)

    def predict(self, test):
        # Predecir usando el clasificador entrenado
        return self.classifier.predict(test)

    def accuracy(self, y, y_pred):
        # Calcular el puntaje de precisión
        return accuracy_score(y, y_pred)

# Función para cargar el archivo .csv
def load_csv(uploaded_file):
    content = uploaded_file['content']
    df = pd.read_csv(io.BytesIO(content))
    return df

def main():
    # Crear un widget de carga de archivos
    upload = FileUpload(accept='.csv', multiple=False)
    display(upload)

    def on_upload_change(change):
        # Verificar si hay archivos cargados
        if upload.value:
            uploaded_file = list(upload.value.values())[0]
            df = load_csv(uploaded_file)

            # Inicializar DataSimulator con los datos cargados
            data_simulator = DataSimulator(df)
            cleaned_data = data_simulator.clean_data()
            features = data_simulator.get_features()

            # Inicializar KMeansProcessor para clustering
            kmeans_processor = KMeansProcessor(features)
            wcss = kmeans_processor.find_optimal_clusters()

            # Inicialización con un número predeterminado de clusters
            initial_clusters = 2
            y_clusters = kmeans_processor.perform_clustering(initial_clusters)

            # Visualización interactiva del gráfico de dispersión
            def update_scatter_plot(num_clusters):
                y_clusters = kmeans_processor.perform_clustering(num_clusters)
                Visualization.scatter_plot(features, y_clusters, kmeans_processor.kmeans_model)
                return y_clusters

            # Mostrar el gráfico de dispersión con el número de clusters predeterminado
            y_clusters = update_scatter_plot(initial_clusters)

            # Widget interactivo para ajustar el número de clusters
            interact(update_scatter_plot, num_clusters=IntSlider(min=1, max=10, step=1, value=initial_clusters))

            # Visualizar el mapa de calor de correlación de los datos limpiados
            Visualization.heatmap(cleaned_data, download=True)

            # Visualizar y guardar el gráfico de dispersión inicial
            Visualization.scatter_plot(features, y_clusters, kmeans_processor.kmeans_model, download=True)

            # Inicializar RandomForestProcessor para clasificación
            random_forest = RandomForestProcessor(features, y_clusters)

            # Datos de prueba para predicción
            test_data = np.array([[5.1, 229, 65.23, 68, 0]])

            # Predecir usando RandomForestClassifier
            prediction = random_forest.predict(test_data)
            accuracy = random_forest.accuracy(y_clusters, random_forest.predict(features))
            print("Prediction:", prediction)
            print("Accuracy:", accuracy)

            # Función para predecir basada en el clasificador
            def give_pred(test):
                prediction = random_forest.predict(test)
                if prediction != 0:
                    return 'Your System Works'
                return 'Your System Failed'

            # Probar la función de predicción
            print(give_pred(test_data))

    # Conectar el evento de cambio de carga de archivos
    upload.observe(on_upload_change, names='value')

# Ejecutar el programa principal
main()
```

Generación de Imágenes con matplotlib:

La función generar_imagen_montanas crea una gráfica simple de montañas utilizando una función seno. La función generar_imagen_atardecer crea una visualización simple de un atardecer en la playa utilizando rectángulos y un círculo para representar el sol.

```python
# Parte 1: Generación de Imágenes Interactivas con Selección de Funciones

import matplotlib.pyplot as plt
import numpy as np
import ipywidgets as widgets
from ipywidgets import interact, interactive, fixed, interact_manual

# Función para generar una imagen de un paisaje con montañas onduladas
def generar_imagen_montanas_onduladas(amplitud=1, frecuencia=1):
    x = np.linspace(-10, 10, 100)
    y = amplitud * np.sin(frecuencia * x)
    plt.plot(x, y)
    plt.fill_between(x, y, color='skyblue', alpha=0.3)
    plt.fill_between(x, -y, color='skyblue', alpha=0.3)
    plt.title("Paisaje con Montañas Onduladas")
    plt.xlabel("Distancia")
    plt.ylabel("Altura")
    plt.show()

# Función para generar una imagen de un paisaje con montañas cuadradas
def generar_imagen_montanas_cuadradas(amplitud=1, frecuencia=1):
    x = np.linspace(-10, 10, 100)
    y = amplitud * np.sign(np.sin(frecuencia * x))
    plt.plot(x, y)
    plt.fill_between(x, y, color='skyblue', alpha=0.3)
    plt.fill_between(x, -y, color='skyblue', alpha=0.3)
    plt.title("Paisaje con Montañas Cuadradas")
    plt.xlabel("Distancia")
    plt.ylabel("Altura")
    plt.show()

# Función para generar una imagen de un atardecer en la playa con un sol redondo
def generar_imagen_atardecer_redondo(sol_y=7, tamano_sol=1):
    fig, ax = plt.subplots()
    ax.set_xlim(0, 10)
    ax.set_ylim(0, 10)
    ax.add_patch(plt.Rectangle((0, 0), 10, 5, color='blue', alpha=0.5))
    ax.add_patch(plt.Rectangle((0, 5), 10, 5, color='orange', alpha=0.5))
    ax.add_patch(plt.Circle((5, sol_y), tamano_sol, color='yellow', alpha=0.8))
    plt.title("Atardecer en la Playa con Sol Redondo")
    plt.show()

# Función para generar una imagen de un atardecer en la playa con un sol cuadrado
def generar_imagen_atardecer_cuadrado(sol_y=7, tamano_sol=1):
    fig, ax = plt.subplots()
    ax.set_xlim(0, 10)
    ax.set_ylim(0, 10)
    ax.add_patch(plt.Rectangle((0, 0), 10, 5, color='blue', alpha=0.5))
    ax.add_patch(plt.Rectangle((0, 5), 10, 5, color='orange', alpha=0.5))
    ax.add_patch(plt.Rectangle((5 - tamano_sol / 2, sol_y - tamano_sol / 2), tamano_sol, tamano_sol, color='yellow', alpha=0.8))
    plt.title("Atardecer en la Playa con Sol Cuadrado")
    plt.show()

# Función para seleccionar y generar imágenes
def seleccionar_generador(tipo_imagen, amplitud=1, frecuencia=1, sol_y=7, tamano_sol=1):
    if tipo_imagen == 'Montañas Onduladas':
        generar_imagen_montanas_onduladas(amplitud, frecuencia)
    elif tipo_imagen == 'Montañas Cuadradas':
        generar_imagen_montanas_cuadradas(amplitud, frecuencia)
    elif tipo_imagen == 'Atardecer con Sol Redondo':
        generar_imagen_atardecer_redondo(sol_y, tamano_sol)
    elif tipo_imagen == 'Atardecer con Sol Cuadrado':
        generar_imagen_atardecer_cuadrado(sol_y, tamano_sol)

# Menú interactivo para seleccionar y generar las imágenes
interact(seleccionar_generador,
         tipo_imagen=['Montañas Onduladas', 'Montañas Cuadradas', 'Atardecer con Sol Redondo', 'Atardecer con Sol Cuadrado'],
         amplitud=(0.1, 2.0, 0.1),
         frecuencia=(0.1, 2.0, 0.1),
         sol_y=(5, 9, 0.1),
         tamano_sol=(0.5, 3.0, 0.1))
```

```python
# Parte 2: Manejo Interactivo de Datos con Pandas e ipywidgets

import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import ipywidgets as widgets
from ipywidgets import interact
from sklearn.linear_model import LinearRegression

# Crear un DataFrame de ejemplo con 10 personas, ciudades de México y una variable binaria de contratación
data = {
    'Nombre': ['Juan', 'María', 'Pedro', 'Ana', 'Luis', 'Sofía', 'Carlos', 'Elena', 'Miguel', 'Laura'],
    'Edad': [28, 24, 35, 42, 30, 27, 33, 26, 29, 31],
    'Ciudad': ['Ciudad de México', 'Guadalajara', 'Monterrey', 'Puebla', 'Tijuana', 'León', 'Mérida', 'Chihuahua', 'Cancún', 'Querétaro'],
    'Último Salario': [3000, 3200, 2800, 3500, 3100, 3300, 2700, 3400, 2900, 3250],
    'Contratado': [1, 0, 1, 1, 0, 1, 0, 1, 0, 1]
}

df = pd.DataFrame(data)

# Función para mostrar el DataFrame y permitir modificaciones en los últimos salarios
def mostrar_dataframe(ultimo_salario_juan=3000, ultimo_salario_maria=3200, ultimo_salario_pedro=2800, ultimo_salario_ana=3500,
                      ultimo_salario_luis=3100, ultimo_salario_sofia=3300, ultimo_salario_carlos=2700, ultimo_salario_elena=3400,
                      ultimo_salario_miguel=2900, ultimo_salario_laura=3250):
    df.loc[df['Nombre'] == 'Juan', 'Último Salario'] = ultimo_salario_juan
    df.loc[df['Nombre'] == 'María', 'Último Salario'] = ultimo_salario_maria
    df.loc[df['Nombre'] == 'Pedro', 'Último Salario'] = ultimo_salario_pedro
    df.loc[df['Nombre'] == 'Ana', 'Último Salario'] = ultimo_salario_ana
    df.loc[df['Nombre'] == 'Luis', 'Último Salario'] = ultimo_salario_luis
    df.loc[df['Nombre'] == 'Sofía', 'Último Salario'] = ultimo_salario_sofia
    df.loc[df['Nombre'] == 'Carlos', 'Último Salario'] = ultimo_salario_carlos
    df.loc[df['Nombre'] == 'Elena', 'Último Salario'] = ultimo_salario_elena
    df.loc[df['Nombre'] == 'Miguel', 'Último Salario'] = ultimo_salario_miguel
    df.loc[df['Nombre'] == 'Laura', 'Último Salario'] = ultimo_salario_laura
    # Mostrar el DataFrame en orden aleatorio para evitar sesgos
    df_aleatorio = df.sample(frac=1).reset_index(drop=True)
    display(df_aleatorio)
    mostrar_dispersión()
    mostrar_heatmap()
    mostrar_resumen()

# Menú interactivo para manejar el DataFrame
interact(mostrar_dataframe,
         ultimo_salario_juan=(2000, 5000, 100), ultimo_salario_maria=(2000, 5000, 100), ultimo_salario_pedro=(2000, 5000, 100),
         ultimo_salario_ana=(2000, 5000, 100), ultimo_salario_luis=(2000, 5000, 100), ultimo_salario_sofia=(2000, 5000, 100),
         ultimo_salario_carlos=(2000, 5000, 100), ultimo_salario_elena=(2000, 5000, 100), ultimo_salario_miguel=(2000, 5000, 100),
         ultimo_salario_laura=(2000, 5000, 100))

# Función para generar un resumen del DataFrame con solo la media y la desviación estándar
def generar_resumen_df(dataframe):
    resumen = dataframe[['Edad', 'Último Salario', 'Contratado']].agg(['mean', 'std']).T
    return resumen

# Función para mostrar el resumen del DataFrame
def mostrar_resumen():
    resumen = generar_resumen_df(df)
    print("Resumen del DataFrame:")
    print(resumen)

# Función para generar un gráfico de dispersión con ajuste lineal y mostrar la ecuación
def mostrar_dispersión():
    plt.figure(figsize=(10, 6))
    sns.regplot(x='Edad', y='Último Salario', data=df)

    # Ajuste lineal
    X = df[['Edad']]
    y = df['Último Salario']
    model = LinearRegression().fit(X, y)
    intercept = model.intercept_
    slope = model.coef_[0]

    # Mostrar la ecuación en el gráfico
    plt.text(0.05, 0.95, f'y = {slope:.2f}x + {intercept:.2f}', ha='left', va='center', transform=plt.gca().transAxes, fontsize=12, bbox=dict(facecolor='white', alpha=0.5))

    plt.title('Dispersión de Últimos Salarios vs Edad con Ajuste Lineal')
    plt.xlabel('Edad')
    plt.ylabel('Último Salario')
    plt.show()

# Función para generar un heatmap de correlaciones
def mostrar_heatmap():
    plt.figure(figsize=(10, 6))
    sns.heatmap(df[['Edad', 'Último Salario', 'Contratado']].corr(), annot=True, cmap='coolwarm', center=0)
    plt.title('Heatmap de Correlaciones')
    plt.show()

# No se requiere botón adicional para actualizar las gráficas, se actualizan automáticamente con el interact
```

```python
# @title Bloque A · Tres divisores: de la temperatura al voltaje y a un archivo CSV
equipo = 0  # @param {type:"slider", min:0, max:10, step:1}
if equipo == 0:
    raise SystemExit("Mueve el deslizador «equipo» a tu número de equipo (1 a 10) y vuelve a ejecutar esta celda.")
rf_mio = RF_EQUIPO[str(equipo)]
divisores = sorted({1000, rf_mio, 100000})          # el de tu equipo y dos de comparación
T = np.arange(0, 50.5, 0.5)
tabla = pd.DataFrame({"T_C": T})
for rf in divisores:
    tabla[f"V_Rf_{rf}"] = np.round(v_div(T, rf), 4)
tabla.to_csv("tres_divisores.csv", index=False)
print(f"Equipo {equipo}: Rf = {rf_mio} Ω. Se guardó tres_divisores.csv con {len(tabla)} renglones.")
display(tabla.head())
for rf in divisores:
    s = (v_div(30, rf) - v_div(20, rf)) / 10 * 1000
    plt.plot(T, v_div(T, rf), lw=3 if rf == rf_mio else 1.5, label=f"Rf = {rf} Ω: {s:.1f} mV/°C a 25 °C")
plt.axvline(25, color="#999", ls=":"); plt.xlabel("Temperatura [°C]"); plt.ylabel("Salida del divisor [V]")
plt.legend(frameon=False); plt.show()
```
